```python
Q1.Explain the difference between Relative , Absolute and Mixed Cell Referencing.
Ans: Relative cell references are the default in Excel. They change automatically when a
formual copied to another cell,maintaining the relative position between cells.There is no
dollar ($) sign in the relative reference for the cell.
*An absolute cell reference in Excel is a way to fix a cell address so that it does not change
when you copy a formula to another cell.we can add dollar($) in front of row and in front of column
to fix the row and column vlaue.
*A mixed cell reference is a type of cell reference in Excel that combines both absolute and relative references.
In a mixed reference, one part of the reference is fixed (absolute) while the other part is allowed to change (relative).
This is achieved by using a dollar sign ($) before the part you want to lock.
```
